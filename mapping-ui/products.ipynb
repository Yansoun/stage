```python
import pandas as pd

products = pd.read_csv("data/raw/northwind/products.csv")

products_df = products.rename(columns={
    "ProductID": "product_id",
    "ProductName": "product_name",
    "SupplierID": "supplier_id",
    "CategoryID": "category_id",
    "UnitPrice": "unit_price",
    "UnitsInStock": "units_in_stock"
})

products_df = products_df[[
    "product_id",
    "product_name",
    "supplier_id",
    "category_id",
    "unit_price",
    "units_in_stock"
]]

products_df.to_csv("data/processed/products.csv", index=False)
```
