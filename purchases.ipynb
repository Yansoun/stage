```python
import pandas as pd
order_details = pd.read_csv("data/order_details.csv")
orders = pd.read_csv("data/orders.csv")
# Parse date ONCE here
orders["OrderDate"] = pd.to_datetime(
    orders["OrderDate"],
    format="%Y-%m-%d %H:%M:%S"
)

purchases = order_details.merge(
    orders[["OrderID", "OrderDate"]],
    on="OrderID",
    how="left"
)

purchases_df = purchases.rename(columns={
    "OrderID": "purchase_id",
    "ProductID": "product_id",
    "Quantity": "quantity",
    "UnitPrice": "unit_price",
    "OrderDate": "purchase_date"
})

# Remove time
purchases_df["purchase_date"] = purchases_df["purchase_date"].dt.date

# Now just remove time
purchases_df["purchase_date"] = purchases_df["purchase_date"].dt.date
```
